## Preprocessing

<span style="color:hotpink;"> As the last step before machine learning, we will proceed with some preprocessing. Checking data types was done in the data overview part of the project and the result showed that the dataset contains numerical variables. Several categorical and binary variables are already numerically encoded. We will proceed with investigation of the value found in obesity and considaterion of scaling of the numerical features, as well as looking at outliers. 

```python
#Looking for outliers using IQR method
#first find the numerical features in the dataset
continous_features=[
    column for column in galle_set.select_dtypes(include='number').columns
    if galle_set[column].nunique() > 5
]
print (continous_features)
```

```python
#check for outliers in continous features  
for column in continous_features:
    Q1 = galle_set[column].quantile(0.25)
    Q3 = galle_set[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = galle_set[(galle_set[column] < lower_bound) | (galle_set[column] > upper_bound)]
    
    print(column, ":", len(outliers), "potential outliers")
```

<span style="color:hotpink;"> For each continous variable, we calculate Q1 and Q3 and use these to find the IQR. Values below Q1 - 1.5 x IQR or above Q3 + 1.5 IQR are identified as potential outliers. It is important to understand potential outliers, as extreme values can be true values. The IQR method identifies unusal observations, but they may be correct. According to the result, some variable such as CRP, ALT, glucose, AST and triglyceride has more than 20 potential outliers. Obesity had one value of 1954, which is one of the 7 potential outliers for this feature. Due to the results finding that BMI could be corresponding to obesity and the calue of 1954 is suspicously high, we put this value to NaN. If the obesity feature is necessary at all, can be considered in the feature selection.

```python
import numpy as np

#replace suspicious values in 'Obesity (%)' with NaN
galle_set.loc[galle_set['Obesity (%)']==1954, 'Obesity (%)'] = np.nan
galle_set['Obesity (%)'].isna().sum()
```
